# How To: Cygwin
Autor: Leon Gjinovci  
Cygwin simuliert eine große Sammlung aus GNU und Open Source Tools um linux-ähnliche Funktionalitäten auf Windows bereitzustellen.

## Installation
Eine Installation wird auf Linux und MacOs nicht benötigt, da diese bereits den Funktionsumfang bieten!

Installationsguide: https://www.cygwin.com/install.html

### Installationsschritte
1. setup-x86_64.exe herunterladen und ausführen
2. Benötigte Packages installieren (z.B. make, git, vim, openssh, wget)

## Verwendung von Cygwin
### Bewegen in Ordnern

Um sich in Cygwin zwischen verschiedenen Ordnern zu bewegen, wird hauptsächlich der Befehl `cd` verwendet. Mit `cd <ordner>` wechselt man in einen bestimmten Unterordner, während `cd ..` in den übergeordneten Ordner zurückführt. Mit `cd ~` gelangt man direkt in das persönliche Home-Verzeichnis. Das aktuell geöffnete Verzeichnis kann jederzeit mit `pwd` angezeigt werden. Mit `ls` lassen sich die Dateien und Unterordner des aktuellen Verzeichnisses auflisten, um sich besser zu orientieren.


### Ordner und Dateien verwalten

Mit `mkdir <name>` erstellt man einen neuen Ordner. Eine leere Datei kann mit `touch <datei>` angelegt werden. Zum Kopieren verwendet man `cp <quelle> <ziel>`; für einen Ordner muss zusätzlich die Option `-r` verwendet werden, zum Beispiel `cp -r <ordner> <ziel>`. Der Befehl `mv <quelle> <ziel>` verschiebt Dateien oder Ordner und kann außerdem zum Umbenennen verwendet werden. Dateien löscht man mit `rm <datei>`, während ein Ordner mitsamt seinem Inhalt mit `rm -r <ordner>` gelöscht wird. Beim Löschen sollte man besonders vorsichtig sein, da die Dateien in Cygwin normalerweise nicht in einen Papierkorb verschoben werden.


### Dateien mit `find` suchen

Mit `find` lassen sich Dateien und Ordner anhand ihres Namens, Typs oder anderer Eigenschaften suchen. Der erste Parameter gibt an, wo die Suche beginnt. `find . -name "*.txt"` sucht beispielsweise ab dem aktuellen Verzeichnis nach allen Textdateien. Mit `find . -type d -name "data"` werden nur Ordner mit dem Namen `data` gefunden. Die Suche kann auch auf einen bestimmten Pfad eingeschraenkt werden, zum Beispiel `find /cygdrive/c/Users -type f -name "*.csv"`. Bei Mustern wie `*.txt` sollte das Muster in Anfuehrungszeichen stehen, damit Cygwin es an `find` uebergibt und nicht bereits die Shell die Dateinamen ersetzt.

Beispiele:

```bash
find . -name "report.txt"
find . -type f -name "*.py"
find . -type d -name "__pycache__"
find . -type f -size +1M
```

### Text mit `grep` suchen

Mit `grep` sucht man nach Textmustern in Dateien. Standardmaessig gibt `grep` die passenden Zeilen zusammen mit dem Dateinamen aus. Das ist hilfreich, um beispielsweise Variablen, Fehlermeldungen oder bestimmte Konfigurationen schnell zu finden.

#### 1. Einfach nach einem Wort suchen

```bash
grep "error" log.txt
```

Gibt alle Zeilen aus `log.txt` aus, die das Wort `error` enthalten. Die Suche unterscheidet zwischen Gross- und Kleinschreibung.

#### 2. Unabhaengig von Gross- und Kleinschreibung suchen

```bash
grep -i "warning" log.txt
```

Die Option `-i` findet sowohl `warning`, `Warning` als auch `WARNING`.

#### 3. Zeilennummern anzeigen

```bash
grep -n "def " script.py
```

Mit `-n` wird vor jeder Trefferzeile die Zeilennummer ausgegeben. So findet man die betreffende Stelle im Quelltext schneller.

#### 4. Nur Dateinamen ausgeben

```bash
grep -l "TODO" *.py
```

Die Option `-l` listet nur Dateien auf, in denen mindestens ein Treffer vorkommt.

#### 5. Dateien rekursiv durchsuchen

```bash
grep -r "password" .
```

Mit `-r` durchsucht `grep` das aktuelle Verzeichnis und alle Unterordner. Fuer Quellcode ist eine Kombination wie `grep -rn "import pandas" .` besonders praktisch.

#### 6. Nur ganze Woerter finden

```bash
grep -w "cat" commands.txt
```

Die Option `-w` verhindert, dass auch Treffer in Woertern wie `concatenate` oder `category` erscheinen.

#### 7. Treffer um den Kontext erweitern

```bash
grep -C 2 "Exception" app.log
```

Mit `-C 2` werden zwei Zeilen vor und nach jedem Treffer angezeigt. Alternativ zeigt `-A 3` drei Folgezeilen und `-B 3` drei vorherige Zeilen.

#### 8. Nach mehreren Mustern suchen

```bash
grep -E "error|warning|failed" app.log
```

Die Option `-E` aktiviert erweiterten regulären Ausdruckssyntax. Der senkrechte Strich `|` bedeutet dabei "oder".

#### 9. Nicht passende Zeilen ausgeben

```bash
grep -v "DEBUG" app.log
```

Mit `-v` werden alle Zeilen ausgegeben, die das Suchmuster nicht enthalten. So kann man beispielsweise Debug-Meldungen aus einer Logdatei ausblenden.

#### 10. Ausgabe von `find` weiterverarbeiten

```bash
find . -type f -name "*.py" -exec grep -nH "TODO" {} \;
```

Dieser Befehl sucht zuerst alle Python-Dateien und fuehrt fuer jede gefundene Datei `grep` aus. `-H` erzwingt die Ausgabe des Dateinamens, `-n` zeigt die Zeilennummer und `{}` steht fuer die jeweils gefundene Datei.


### Kleines `grep`-Lexikon

| Begriff/Option | Bedeutung |
| :--- | :--- |
| Muster | Der gesuchte Text oder regulaere Ausdruck. |
| `-i` | Ignoriert Gross- und Kleinschreibung. |
| `-v` | Gibt Zeilen ohne Treffer aus. |
| `-n` | Zeigt die Zeilennummern an. |
| `-r` | Durchsucht Verzeichnisse rekursiv. |
| `-l` | Gibt nur die Namen von Dateien mit Treffern aus. |
| `-L` | Gibt nur die Namen von Dateien ohne Treffer aus. |
| `-c` | Zaehlt die Treffer pro Datei. |
| `-w` | Sucht nur nach ganzen Woertern. |
| `-x` | Verlangt, dass die gesamte Zeile dem Muster entspricht. |
| `-nH` | Zeigt Zeilennummer und Dateinamen an. |
| `-A N` | Zeigt `N` Zeilen nach dem Treffer. |
| `-B N` | Zeigt `N` Zeilen vor dem Treffer. |
| `-C N` | Zeigt `N` Zeilen Kontext vor und nach dem Treffer. |
| `-E` | Aktiviert erweiterte regulaere Ausdruecke, zum Beispiel `error|warning`. |
| `-F` | Behandelt das Muster als festen Text statt als regulaeren Ausdruck. |
| `--include="*.py"` | Durchsucht rekursiv nur Dateien mit passender Endung. |
| `--exclude="*.log"` | Ueberspringt Dateien mit passender Endung. |
| `-q` | Gibt keine Treffer aus; nur der Rueckgabestatus wird verwendet. |
| `\|` | Verbindet Befehle ueber eine Pipeline, zum Beispiel `cat app.log \| grep error`. |

Ein praktischer Alltagsbefehl ist zum Beispiel `grep -Rni --include="*.py" "todo" .`. Er durchsucht alle Python-Dateien unterhalb des aktuellen Verzeichnisses, ignoriert die Gross- und Kleinschreibung und zeigt Dateinamen sowie Zeilennummern an.

## Cheatsheet

| Befehl | Beschreibung |
| :--- | :--- |
| `pwd` | Zeigt das aktuelle Arbeitsverzeichnis an. |
| `ls` | Listet Dateien und Ordner im aktuellen Verzeichnis auf. |
| `ls -la` | Listet auch versteckte Dateien mit Details auf. |
| `cd <ordner>` | Wechselt in den angegebenen Ordner. |
| `cd ..` | Wechselt in das übergeordnete Verzeichnis. |
| `cd ~` | Wechselt in das Home-Verzeichnis. |
| `mkdir <name>` | Erstellt einen neuen Ordner. |
| `touch <datei>` | Erstellt eine leere Datei. |
| `cp <quelle> <ziel>` | Kopiert eine Datei oder einen Ordner. |
| `mv <quelle> <ziel>` | Verschiebt oder benennt eine Datei bzw. einen Ordner um. |
| `rm <datei>` | Löscht eine Datei. |
| `rm -r <ordner>` | Löscht einen Ordner inklusive seines Inhalts. Vorsichtig verwenden. |
| `cat <datei>` | Gibt den Inhalt einer Datei im Terminal aus. |
| `grep <muster> <datei>` | Sucht nach einem Textmuster in einer Datei. |
| `find . -name "<muster>"` | Sucht ab dem aktuellen Verzeichnis nach passenden Dateien. |
| `clear` | Löscht die Anzeige im Terminal. |
| `man <befehl>` | Öffnet die Handbuchseite eines Befehls. |
| `echo <text>` | Gibt Text im Terminal aus. |
| `chmod +x <datei>` | Macht eine Datei ausführbar. |
| `cd /cygdrive/c` | Wechselt zum Windows-Laufwerk `C:\`. |
